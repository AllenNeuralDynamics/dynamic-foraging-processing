This notebook demonstrates how to build the `trials` table from a raw dynamic foraging acquisition using `TrialTableBuilder`. Point a `RawDataLoader` at a dataset acquired via the `aind-behavior-dynamic-foraging` [data contract](https://github.com/AllenNeuralDynamics/Aind.Behavior.DynamicForaging/blob/main/src/aind_behavior_dynamic_foraging/data_contract/_dataset.py), then pass `loader.dataset` to the builder.

```python
%load_ext autoreload
%autoreload 2
from pathlib import Path

from dynamic_foraging_processing.raw_data_loader import RawDataLoader
```

## Create a loader

Point `RawDataLoader` at the root directory of a dataset acquisition.

```python
# Update this path to point at a real acquisition directory on your machine.
dataset_path = Path(
    r"C:\Users\arjun.sridhar\Downloads\821586_2026-04-28T200347Z\821586_2026-04-28T200347Z"
)

loader = RawDataLoader(path=dataset_path)
loader
```

## Build the trials table

Pass `loader.dataset` to `TrialTableBuilder` and call `build()` to get the trials table as a `pandas.DataFrame` (one row per trial).

```python
from dynamic_foraging_processing.processing import TrialTableBuilder

builder = TrialTableBuilder(loader.dataset)
trial_table = builder.build()
trial_table.head()
```

## What's in the table

Each row is one trial; each column maps to a field on the `TrialConfig` Pydantic model in
[`trial_config.py`](../src/dynamic_foraging_processing/processing/models/trial_config.py). That
file is the source of truth for the column set and the per-column descriptions (which also become
the NWB trial-column descriptions). The columns, grouped as they appear in the model:

- **Trial timing:** `start_time`, `stop_time`
- **Trial info:** `animal_response` (0=left, 1=right, 2=no response), `rewarded_historyL`,
  `rewarded_historyR`, `delay_start_time`, `goCue_start_time`
- **Behavior structure:** `bait_left`, `bait_right`, `base_reward_probability_sum`,
  `reward_probabilityL`, `reward_probabilityR`, `left_valve_open_time`, `right_valve_open_time`
- **Block information:** `block_beta`, `block_min`, `block_max`, `min_reward_each_block`
- **Delay duration:** `delay_beta`, `delay_min`, `delay_max`, `delay_duration`
- **ITI duration:** `ITI_beta`, `ITI_min`, `ITI_max`, `ITI_duration`
- **Response & reward duration:** `response_duration`, `reward_consumption_duration`
- **Autowater:** `auto_waterL`, `auto_waterR`
- **Lickspout position:** `lickspout_position_x`, `lickspout_position_y1`,
  `lickspout_position_y2`, `lickspout_position_z`

To see the description for every column without leaving the notebook, call
`TrialConfig.column_descriptions()` (cell below).
