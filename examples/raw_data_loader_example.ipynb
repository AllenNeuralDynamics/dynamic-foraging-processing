# RawDataLoader Example

This notebook demonstrates how to use the `RawDataLoader` from `dynamic_foraging_processing` to load raw dynamic foraging data via the `aind-behavior-dynamic-foraging` [data contract]( https://github.com/AllenNeuralDynamics/Aind.Behavior.DynamicForaging/blob/main/src/aind_behavior_dynamic_foraging/data_contract/_dataset.py). Data must be acquired in this format to be compatible with these tools.


```python
%load_ext autoreload
%autoreload 2
from pathlib import Path

from dynamic_foraging_processing.raw_data_loader import RawDataLoader
```

## Create a loader

Point `RawDataLoader` at the root directory of a dataset acquisition.


```python
# Update this path to point at a real acquisition directory on your machine.
dataset_path = Path(
    r"C:\Users\arjun.sridhar\Downloads\821586_2026-04-28T200347Z\821586_2026-04-28T200347Z"
)

loader = RawDataLoader(path=dataset_path)
loader
```

## Access an individual stream

You can load a single stream directly from `loader.dataset` without loading every stream in the dataset. See the [data contract]( https://github.com/AllenNeuralDynamics/Aind.Behavior.DynamicForaging/blob/main/src/aind_behavior_dynamic_foraging/data_contract/_dataset.py) on how to access the various streams via the tree-like contract. An example is shown below


```python
stream = loader.dataset.at("Behavior").at("PreviousMetrics").load().data
stream
```

## Load all raw data

`get_all_raw_data` returns a dictionary mapping each stream  to its loaded data (a `pandas.DataFrame` or a `dictionary`). 

Note: The keys returned chain the parent stream and the current stream, since some **child leaf node streams have the same name but come from different parents**. This dictionary here will then be used to create the **acquisition container in the NWB file**

Note: Version mismatch warnings below are expected when loading datasets acquired with an **older version of the data contract**. Data will still load correctly via best-effort coercion.


```python
raw_data = loader.get_all_raw_data()
list(raw_data.keys())
```

```python
# Inspect same stream from fully loaded raw data
stream_name = next(iter(raw_data))
print(f"Looking at stream {stream_name}:")
raw_data[stream_name]
```

```python
# Inspect another stream
stream_name = "InputSchemas.TaskLogic"
print(f"Looking at stream {stream_name}:")
raw_data[stream_name].model_dump()
```
